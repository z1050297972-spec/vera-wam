# Part 0:环境安装
在WSL下逐步执行

## 1) WSL2(Windows PowerShell 管理员):wsl --install -d Ubuntu-22.04,重启后进去
## 2) 驱动确认(WSL 内):nvidia-smi 应能看到两张 3090

## 3) 下载安装包,并确认下载成功(应显示约 140 MB 的文件)
wget https://repo.anaconda.com/miniconda/Miniconda3-latest-Linux-x86_64.sh
ls -lh Miniconda3-latest-Linux-x86_64.sh

## 4) 静默安装(-b 批处理模式,-p 指定装到家目录下)
bash Miniconda3-latest-Linux-x86_64.sh -b -p $HOME/miniconda3

## 5) 把 conda 注册进 bash(以后开终端直接能用 conda 命令)
$HOME/miniconda3/bin/conda init bash   

exec bash         #重载 shell;或者直接关掉终端重开一个
 
## 6) 建环境并激活
如果下载了docker，请在进入执行 wsl --setdefault Ubuntu-22.04 ，将默认wsl设置为我们下载的


conda create -n vera python=3.11 -y


conda activate vera

## 7) 安装pytorch 2.6 + CUDA 12.4
pip install torch==2.6.0 torchvision --index-url https://download.pytorch.org/whl/cu124

## 8) VERA本体(把路径换成你的挂载点,如 /mnt/c/Users/king OF LOL/Desktop/vera_wam/vera-main)
wslpath "C:\Users\king  OF LOL\Desktop\vera_wam\vera-main"    输出:/mnt/c/Users/king  OF LOL/Desktop/vera_wam/vera-main
cd "$(wslpath "C:\Users\king  OF LOL\Desktop\vera_wam\vera-main")"
/mnt/c/Users/king  OF LOL/Desktop/vera_wam/vera-main
